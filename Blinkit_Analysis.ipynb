```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("blinkit_orders.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["delivery_status"].value_counts()
```

```python
import matplotlib.pyplot as plt

df["delivery_status"].value_counts().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Delivery Status")
plt.xlabel("Status")
plt.ylabel("Number of Orders")

plt.show()
```

```python
df["payment_method"].value_counts()
```

```python
df["payment_method"].value_counts().plot(
    kind="bar",
    figsize=(8,5),
    color="orange"
)

plt.title("Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["order_total"], bins=20)

plt.title("Order Total Distribution")
plt.xlabel("Order Total")
plt.ylabel("Number of Orders")

plt.show()
```

```python
df["delivery_status"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6)
)

plt.title("Delivery Status Distribution")
plt.ylabel("")

plt.show()
```
